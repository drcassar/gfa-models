# Imports

```python
import shap
import matplotlib.pyplot as plt
from joblib import dump, load
```

# SHAP Analysis
## CHEM Model

```python
# loading SHAP explanation
explanation_CHEM = load('../data/support/explanation_shap/explanation_shap_CHEM.joblib')

shap_values_CHEM = explanation_CHEM.values
print(f"shap_values_CHEM shape: {shap_values_CHEM.shape}")
print(f"explanation_CHEM shape: {explanation_CHEM.shape}")

explanation_glass_CHEM = explanation_CHEM[:,:,1]
shap_values_glass_CHEM = shap_values_CHEM[:, :, 1]
```

### Violin Plot

```python
# Summary plot
shap.summary_plot(explanation_glass_CHEM, max_display=70, plot_type="violin", show=False)

# Save the figure
plt.tight_layout() 
# plt.savefig('shap_[name]_violin_plot.png', dpi=300, bbox_inches='tight')
# plt.savefig('shap_[name]_violin_plot.pdf', bbox_inches='tight')
# plt.savefig('shap_[name]_violin_plot.svg', bbox_inches='tight')
plt.show()
```

### Bar Plot

```python
# Summary plot
#shap.summary_plot(explanation_glass, max_display=20, plot_type="bar", show=False)
shap.summary_plot(explanation_glass_CHEM, plot_size=(10, 10), max_display=20, plot_type="bar", show=False)

# Grid behind the plot
ax = plt.gca()
ax.set_axisbelow(True)
ax.grid(True, axis='x')#, linewidth=1, alpha=0.5)

# Save the figure
plt.tight_layout() 
# plt.savefig('shap_[name]_bar_plot.png', dpi=300, bbox_inches='tight')
# plt.savefig('shap_[name]_bar_plot.pdf', bbox_inches='tight')
# plt.savefig('shap_[name]_bar_plot.svg', bbox_inches='tight')
plt.show()
```

```python
explanation_glass_CHEM.feature_names
```

### Others

```python
shap.plots.scatter(explanation_glass_CHEM[:, "Ge"], color=explanation_glass_CHEM[:, "O"])
```
