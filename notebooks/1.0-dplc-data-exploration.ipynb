# Introduction 

<p style="text-align: justify"> In this jupyter notebook, the main objective is to explore the data from a glass formation dataset initially obtained by the <i> SciGlass Python</i> module, having been pre-processed by Daniel Cassar, and to make the appropriate changes for the use of machine learning methods on the dataset. This dataset initially contains more than 72k examples of glasses, crystals and intermediates compositions, in addition to 319 attributes associated with these materials. There are 317 quantitative attributes of inorganic compounds that may be present in the material, as well as 1 quantitative attribute of a unique code associated with each material (<code>Kod</code>) and 1 qualitative attribute about if the compounds form glass, crystals or intermediates (<code>GF</code>).

### Imported Required Librarie

```python
import pandas as pd
```

## Exploratory Data Analytics

### Basic Exploratory Data Analytics and Basic Processing 

<p style="text-align: justify"> This subsection shows a brief analysis of the dataset, with a descriptive analysis, exclusion of duplicate examples and a better ordering of the data. Firstly, the file was imported from inside a zip folder and partially displayed to briefly observe it <sup>[1, 2, 3]</sup>.

```python
df = pd.read_csv("../data/raw/GF_data_compound.zip", sep = "\t")
display(df)
```


<p style="text-align: justify"> The quantitative attribute of a unique code for each example will be deleted because of the non-relevance in machine learning models. It will also check how many missing and negative elements there are, as well as how many duplicate examples or duplicate attributes there are. In addition, since the attributes are a percentage of the composition of the solid studied, it will be checked how many examples there are where the sum of the attributes doesn't result in 100% <sup>[3, 4, 5, 6]</sup>.

```python
df = df.drop('Kod', axis=1) # drop the code attribute

print('Number of missing values:', df.isnull().sum().sum())
print('Number of negative values:', (((df.drop('GF', axis=1)) < 0).sum().sum()))

print('''Number of duplicated attributes:''', (df.columns.duplicated().sum()))
print('Number of duplicated examples:', df.duplicated().sum())
print('''Number of examples that the sum of attributes doesn't result in 100%:''', (round((df.drop('GF', axis=1)).sum(axis=1), 5) != 100).sum())
```

<p style="text-align: justify"> There are negative elements and duplicate examples, the duplicate examples are deleted. In addition, the number of examples with at least one negative value was checked and descriptive statistics of the dataset were displayed to better understand the negative values <sup>[3, 4, 5, 6]</sup>.


```python
df = df.drop_duplicates(keep=False) # drop_duplicates

print('Number of examples with at least one negative value:', len(df[(df.select_dtypes(include='number') < 0).any(axis=1)]))

pd.set_option('display.max_rows', None)  # show all rows 
pd.set_option('display.max_columns', None) # show all columns

round(df.describe(), 2) # show descriptive statistics of the dataset
```

<p style="text-align: justify"> It's important to understand that negative values are errors, it's impossible to have negative values as a composition. In this sense, the presence of these values is probably human error, these values should most likely be zero. Therefore, the negative numbers will be set to zero and the rows that had negative values will have all the values normalized so that the sum of the attributes results in 100%. After this, it will be checked if there are still negative values and if the sum of the attributes of any row does not result in 100% <sup>[3, 7, 8]</sup>.

```python
categorical_column = df['GF'] # The GF column will be temporarily separated.

df = df.drop('GF', axis=1) # drop "GF" column 

df[df < 0] = 0 # negative numbers are equal to zero

df = df.div(df.sum(axis=1), axis=0) * 100 # normalize the values ​​of each row

print('Number of negative values:', ((df < 0).sum().sum()))

print('''Number of examples that the sum of attributes doesn't result in 100%:''', (round(df.sum(axis=1), 5) != 100).sum())

df = pd.concat([df, categorical_column], axis=1) # add the 'GF' column back to the dataset
```

<p style="text-align: justify"> Finally, the index will be reset and the size and descriptive statistics of the current dataset will be displayed <sup>[1, 2, 9]</sup>.

```python
df = df.reset_index(drop=True)

print('Dimensionality of the DataFrame:', df.shape) # show the dimensionality of the DataFrame
round(df.describe(), 2)
```

## Analysis of Numeric Attributes and <code>GF</code> Labels, and Sorting the Dataset

### Basic Analysis of Numeric Attributes and <code>GF</code> Labels

<p style="text-align: justify"> All the numerical attributes of the dataset were show:

```python
for column in df.columns:
    print(column)
```

<p style="text-align: justify"> All the <code>GF</code> labels and the number of times they are present in the dataset have been displayed. There are 209 labels in total. <sup>[10]</sup>

```python
display(df['GF'].value_counts())

pd.reset_option('display.max_rows')
pd.reset_option('display.max_columns')
```

### Sorting the Dataset

<p style="text-align: justify"> Finally, the dataset was sorted based on the <code>GF</code> attribute, using as a criterion the number of times each label is present in the dataset in descending order. <sup>[10]</sup>

```python
df = df.iloc[df.groupby('GF').GF.transform('size').mul(-1).argsort(kind='mergesort',)].reset_index(drop=True)
df = df.reset_index(drop=True)
display(df)
```

## Selecting GF Labels

<p style="text-align: justify"> In the previous topic, you can see that there are 209 <code>GF</code> labels about what the composition forms: glass, crystal or an intermediate. To simplify things in a way that is probably more useful for applying machine learning models, the labels will be reduced to just three: crystal (<code>value -1</code>), glass (<code>value 1</code>) and intermediates (<code>value 0</code>). To do this, labels that appear at least 100 times will be analyzed and either classified into one of the three labels mentioned or just deleted. <sup>[11]</sup>

```python
glass = ['Glass','Clear glass', 'Opaque glass', 'Partially opaque glass', 'Phase-separated glass', 'Stable glass', 'Opal glass', 'Phase separated glass', 
        'Phase-separation', 'Homogeneous glass', 'Heterogeneous glass', 'Opaque']
crystal = ['Crystal', 'Crystalline', 'Crystallization']
glass_and_crystal = ['Glass and crystal', 'Glass+crystal', 'Glass containing a crystalline part', 'Glass + crystal', 'Clear glass with crystalline part', 'Glass with crystal']
delete = ['Imperfect melt', 'Unmelted', 'Batch stone', 'Crystal or imperfect melt', 'Surface crystallization', 'Fully devitrified glass', 'Partial divitrification or opal glasses']

labels  = glass + crystal + glass_and_crystal

df = df[df['GF'].isin(labels)] 
df = df.replace(glass, 1).replace(crystal, -1).replace(glass_and_crystal, 0)
```

Check how many times the labels appear.<sup>[10]</sup>

```python
display(df['GF'].value_counts())
```

<p style="text-align: justify"> The dimensionality and the descriptive statistics of the current dataset will be displayed <sup>[1, 2, 9]</sup>.

```python
print('Dimensionality of the DataFrame:', df.shape) # show the dimensionality of the DataFrame
round(df.describe(), 2)
```

Finally, the dataset was saved as a <code>.csv</code> file.

```python
df.to_csv("../data/processed/GFdata_compound_processed.csv", index=False)
```
