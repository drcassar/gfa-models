## Mean

```python
sns.set_style('darkgrid')

fig, ax = plt.subplots(figsize=(10, 6))

# Plot the mean F1-score across folds as a line
sns.lineplot(
    data=df_sfs_results, 
    x='num_features', 
    y='mean_score', 
    ax=ax, 
    color='b', 
    label='Mean F1-Score'
)

# Compute the 95% confidence interval around the mean
# Formula: CI = mean ± 1.96 * (std / sqrt(n))
# - std = standard deviation of the folds
# - n = number of folds (k=10 in this case)
n_folds = 10
ci = 1.96 * df_sfs_results['std_score'] / np.sqrt(n_folds)

# Add a shaded area for the confidence interval
ax.fill_between(
    df_sfs_results['num_features'],
    df_sfs_results['mean_score'] - ci,
    df_sfs_results['mean_score'] + ci,
    color='b',
    alpha=0.2,
    label='95% CI'
)

# Labels and formatting
ax.set_title('F1-Score Macro (Mean) vs Number of Features Selected', fontsize=16)
ax.set_xlabel('Number of Features Selected', fontsize=14)
ax.set_ylabel('F1-Score Macro (Mean)', fontsize=14)
ax.set_ylim(0.6, 0.8)

plt.legend()
plt.tight_layout()
plt.grid(lw=1.5)
plt.show()

sns.reset_defaults()
```

## Standard Deviation (std)

```python
sns.set_style('darkgrid')

fig, ax = plt.subplots(figsize=(10, 6))
sns.lineplot(data=df_sfs_results, x='num_features', y='std_score', ax=ax, color='b')#'skyblue')
ax.set_title('Variation of Standard Deviation vs Number of Selected Features', fontsize=16)
ax.set_xlabel('Number of Features Selected', fontsize=14)
ax.set_ylabel('Standard Deviation of F1-Score', fontsize=14)
#ax.set_ylim(0.6, 0.8)
#ax.set_xlim(0, 20)
#ax.set_xticks(range(0, 21, 1))
#ax.set_xticklabels(df_sfs_results['num_features'])
plt.tight_layout()
plt.grid(lw=1.5)
plt.show()

sns.reset_defaults()
```

```python
df_sfs_results['std_smooth'] = df_sfs_results['std_score'].rolling(window=10, center=True, min_periods=1).mean()

sns.set_style('darkgrid')

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(df_sfs_results['num_features'], df_sfs_results['std_smooth'], color='b', label='Smoothed Std')
ax.scatter(df_sfs_results['num_features'], df_sfs_results['std_score'], color='b', alpha=0.2, label='Original Std')
ax.set_title('Smoothed Standard Deviation vs Number of Features', fontsize=16)
ax.set_xlabel('Number of Features Selected', fontsize=14)
ax.set_ylabel('Standard Deviation of F1-Score', fontsize=14)
ax.legend()
plt.tight_layout()
plt.show()
sns.reset_defaults()
```
